```python
import cv2
import numpy as np
import json
from tensorflow.keras.models import load_model
from collections import deque, Counter

# Load trained model + label map
MODEL_PATH = 'model/sign_language_model.h5'
LABELS_PATH = 'model/labels.json'
IMG_SIZE = 64

print(" Loading model...")
model = load_model(MODEL_PATH)
print("Model loaded successfully!")

with open(LABELS_PATH, 'r') as f:
    labels = json.load(f)

print(f" Loaded {len(labels)} labels: {labels}")

#  Start Webcam
cap = cv2.VideoCapture(0)
if not cap.isOpened():
    print(" Error: Could not access webcam.")
    exit()

current_text = ""
prediction_buffer = deque(maxlen=15) # Buffer for smoothing predictions

print("Press 'Q' to quit the live window.")
print("Press 'Space' to insert a space.")

# Real-time Prediction Loop
while True:
    ret, frame = cap.read()
    if not ret:
        print("Failed to read from webcam.")
        break

    # Define region of interest (ROI)
    x1, y1, x2, y2 = 100, 100, 300, 300
    cv2.rectangle(frame, (x1, y1), (x2, y2), (255, 0, 0), 2) # Draw Blue ROI

    roi = frame[y1:y2, x1:x2]

    # FIX: Ensure RGB conversion for consistency with training
    img = cv2.cvtColor(roi, cv2.COLOR_BGR2RGB) 

    # Preprocess ROI for model
    img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
    img = img.astype("float32") / 255.0
    img = np.expand_dims(img, axis=0)

    # Predict
    prediction = model.predict(img, verbose=0)
    confidence = np.max(prediction)
    class_index = np.argmax(prediction)
    predicted_label = labels[class_index]

    # Use buffer for stable prediction
    if confidence > 0.85:
        prediction_buffer.append(predicted_label)
    
    # Majority vote to trigger text addition
    final_label = ""
    if len(prediction_buffer) > 10: 
        most_common = Counter(prediction_buffer).most_common(1)
        if most_common:
            final_label, count = most_common[0]
            
            # Require the label to be the majority
            if count >= prediction_buffer.maxlen * 0.7: # e.g., 11 out of 15
                 # Check if the last added character is the same to prevent rapid duplicates
                if not current_text or final_label != current_text[-1]:
                    current_text += final_label
                prediction_buffer.clear() # Reset buffer after adding character

    # Draw Prediction Info
    cv2.putText(frame, f"Pred: {predicted_label} ({confidence*100:.1f}%)",
                (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)
    cv2.putText(frame, f"TEXT: {current_text}",
                (50, 400), cv2.FONT_HERSHEY_SIMPLEX, 1.0, (255, 255, 255), 2)
    
    # Show window
    cv2.imshow("SignSpeak AI - Live", frame)

    key = cv2.waitKey(1) & 0xFF
    if key == ord('q'):
        break
    elif key == ord(' '): # Add space key
        current_text += " "
        prediction_buffer.clear()

# Cleanup
cap.release()
cv2.destroyAllWindows()
print("\n Session ended.")
```
